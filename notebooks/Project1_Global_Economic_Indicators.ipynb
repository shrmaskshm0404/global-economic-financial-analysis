# Project 1 — Global Economic & Financial Indicators

SQL + Python portfolio project using a 2015–2024 multi-country macroeconomic dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'global_economic_indicators.csv')
df.head()
```

## 1. Data quality checks

```python
print(df.shape)
print(df.isna().sum())
print('Duplicate country-year rows:', df.duplicated(['country','year']).sum())
```

## 2. Latest snapshot

```python
latest = df[df.year == df.year.max()].sort_values('gdp_current_usd_bn', ascending=False)
latest[['country','gdp_current_usd_bn','inflation_pct','unemployment_pct','trade_pct_gdp']]
```

## 3. GDP growth

```python
df = df.sort_values(['country','year'])
df['gdp_yoy_pct'] = df.groupby('country')['gdp_current_usd_bn'].pct_change()*100
df[['country','year','gdp_yoy_pct']].dropna().head(20)
```

## 4. GDP trend

```python
plt.figure(figsize=(10,6))
for country, g in df.groupby('country'):
    plt.plot(g.year, g.gdp_current_usd_bn, label=country)
plt.title('GDP Trend by Country')
plt.xlabel('Year'); plt.ylabel('GDP (USD billions)'); plt.legend(ncol=2, fontsize=8); plt.show()
```

## 5. Inflation trend

```python
plt.figure(figsize=(10,6))
for country, g in df.groupby('country'):
    plt.plot(g.year, g.inflation_pct, label=country)
plt.title('Inflation Trend by Country')
plt.xlabel('Year'); plt.ylabel('Inflation (%)'); plt.legend(ncol=2, fontsize=8); plt.show()
```

## 6. 2024 unemployment ranking

```python
u = latest.sort_values('unemployment_pct')
plt.figure(figsize=(9,5))
plt.barh(u.country, u.unemployment_pct)
plt.title('Unemployment Rate — 2024')
plt.xlabel('Unemployment (%)'); plt.show()
```

## 7. Inflation vs unemployment

```python
plt.figure(figsize=(8,6))
plt.scatter(latest.inflation_pct, latest.unemployment_pct)
for _, r in latest.iterrows():
    plt.annotate(r.country, (r.inflation_pct, r.unemployment_pct), xytext=(5,5), textcoords='offset points', fontsize=8)
plt.title('Inflation vs Unemployment — 2024')
plt.xlabel('Inflation (%)'); plt.ylabel('Unemployment (%)'); plt.show()
```
